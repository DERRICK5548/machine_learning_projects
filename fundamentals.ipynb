# Variable length arguments
*args (variable position arguments) -> where you can pass a variable number of arguments


````python
def sum_all (*numbers):
    total = 0
    for num in numbers:
        total += num
    return total
print(sum_all(1,2,3))
print(sum_all(1,2,3,4,5))
````

**kwargs-> Variable number of keyword arguments


````python
def print_info(**info):
    for key, value in info.items():
        print(f'{key}: {value}')

print_info(age=21, name='Grace', job= 'Developer', city='Mombasa')
print_info()
````

# Error Handling (Syntax)

try:
   block of code that might cause error
   risky_code()
except:
    code that runs if error occurs
    print('An error occured")


````python
# Try except else finally
try:
     number = int(input("Enter a number: "))
     result = 10/ number
     print("We have gone through try")
except ZeroDivisionError:
    print ("Zero division error, please try again")
except ValueError:
    print("Invalid number")
except Exception as e:
    print(f"An error occured: {e}")
else:
     print(f"The result is {result}")
finally:
    print("Thanks for your time")
````
