## Unsupervised Learning and Clustering

### K-means Clustering
Its unsupervised machine learning algorithim-
It partitions 'unlabeled data' into K distinct groups known as (clusters)

#### What is K?
- This is the number of groups defined.


## What is unsupervised learning?
### Supervised vs unsupervised

| Supervised | Unsupervised|
|--- | --- |
|Training data has labels | Training data has no labels|
|Goal: Learn a mapping from inputs to outputs | Goal: Find hidden structure and patterns in the data.|
| Example use cases: Spam detection, house prices prediction, cancer diagnosis| Examples: Customer segmentation, anomaly detection.|

# K-means clustering
## How K-means works?
- It works by:
1. Iteratively assigning data points to the nearest cluster centroid—using Euclidean distance
2. and recalculating centroids to minimize within-cluster variance


## The elbow method -> How do we choose K?
- By K we mean the centroid
- Select a range of k values (eg 1- 10)



```python
import numpy as np
np.random.seed(42)
```

```python
import pandas as pd
n = 300
customers = pd.DataFrame({
'annual_spend': np.r_[
np.random.normal(2000, 400, 75), # Low spenders
np.random.normal(5000, 600, 75), # Mid spenders
np.random.normal(9000, 800, 75), # High spenders
np.random.normal(3500, 500, 75), # Occasional spenders
],
'visit_frequency': np.r_[
np.random.normal(5, 2, 75),
np.random.normal(20, 4, 75),
np.random.normal(35, 5, 75),
np.random.normal(10, 3, 75),
],
'avg_basket_size': np.r_[
np.random.normal(100, 20, 75),
np.random.normal(250, 40, 75),
np.random.normal(400, 60, 75),
np.random.normal(180, 30, 75),
]
})
```

```python
customers.describe()
```

```python
customers.info()
```

```python
customers.head()
```

```python
# scaling the dataset
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(customers)
```

```python
# Elbow method -> Finding the best number of clusters (K)
from sklearn.cluster import KMeans
inertias = []

for k in range(1, 11):
    km = KMeans(n_clusters=k , random_state=42, n_init=10)
    km.fit(X_scaled)
    inertias.append(km.inertia_)
    
```

```python
# Plot to pick the best K
import matplotlib.pyplot as plt

plt.figure(figsize=(8 ,4))
plt.plot(range(1,11), inertias, 'o--' , linewidth = 2, markersize='8' )
plt.title('Elbow method -Finding the optimal K')
plt.xlabel('Number of clusters K')
plt.ylabel('Inertia (Within-Sum Cluster Sum of Squares)')
plt.xticks(range(1, 11))
plt.grid(alpha=0.3)
plt.show()
```

```python
# Train the model based on the best K
km_final = KMeans(n_clusters=4, random_state=42,n_init=10)
customers['cluster'] = km_final.fit_predict(X_scaled)
customers
```

```python
# profile for each cluster
profile = customers.groupby('cluster').agg({
    'annual_spend': 'mean',
    'visit_frequency': 'mean',
    'avg_basket_size': 'mean',
    'cluster': 'count'
}).rename(columns ={'cluster': 'count'}).round(0)

print('cluster profiles')
print(profile)
```

```python
from cProfile import label


colours = ['#3498db',  '#2ecc71', '#e74c3c', '#f39c12']
labels = ['Mid_value','VIP customers', 'Low spenders', 'Occasional']

plt.figure(figsize=(9 ,5))
for c in range(4):
    mask = customers['cluster'] == c
    plt.scatter(
        customers[mask]['annual_spend'],
        customers[mask]['visit_frequency'],
        c=colours[c],
        label = labels [c],
        alpha = 0.7,
        s = 50
    )
# Mark centroids
centroids_orig = scaler.inverse_transform(km_final.cluster_centers_)
plt.scatter(
    centroids_orig[:, 0],
    centroids_orig[:, 1],
    s = 200,
    marker='x',
    c='black', 
    zorder=5, 
    label='Centroids'
    )
plt.xlabel('Annual Spend (KES)'),
plt.ylabel('Visit Frequency (per month)'),
plt.title('Customer Segments - K-Means (K=4)'),
plt.legend(),
plt.grid(alpha = 0.2)
```

```python
# Hierarchial  Clustering

# Picking a subset of 60 customers for readability
import numpy as np

subset_ids = np.random.choice(len(X_scaled), 60, replace = False )
X_sub = X_scaled[subset_ids]
```

```python
subset_ids
```

```python
# Linkage
from scipy.cluster.hierarchy import dendrogram, linkage

z = linkage(X_sub, method='ward')
```

```python
# Plot dendogram
import matplotlib.pyplot as plt

plt.figure(figsize = (14, 5))
dendrogram(z, truncate_mode='level')
plt.axhline(y=6, color='red', linestyle ='--', linewidth=1.5, label='cut here for 4 clusters')
plt.axhline(y=3, color='blue', linestyle =':', linewidth=1.5, label='cut here for 4 clusters')
plt.legend()
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(n_clusters=4, linkage='ward')
customers['hc_cluster'] = hc.fit_predict(X_scaled)
customers
```

```python
print("K-means cluster profiles")
print(customers.groupby('cluster')[['annual_spend', 'visit_frequency']].mean().round(0))
```

```python
print("Hierarchial cluster profiles")
print(customers.groupby('hc_cluster')[['annual_spend','visit_frequency']].mean().round(0))
```

```python
# side by side scatter plot
fig, axes = plt.subplots(1,2, figsize=(14,5))
# The loop picks what we want to plot and the inner loop plots the actual diagram.
for ax, col, titleX in zip(
    axes,
    ['cluster', 'hc_cluster'],
    ['K-means cluster', 'Hierarchial cluster']
):
  for c in range(4):
      mask = customers[col] == c
      ax.scatter(customers[mask]['annual_spend'], customers[mask]['visit_frequency'],
                alpha=0.7, s=40,label=f'cluster {c}')
      ax.set_xlabel('annual_spend')
      ax.set_ylabel('Visit frequency')
      ax.set_title(titleX)
      ax.legend()
      ax.grid(alpha = 0.2)
plt.tight_layout()
plt.show()      
```
